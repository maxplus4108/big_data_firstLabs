## Задание 2. Влияние формы обучения на активность
Решите гипотезу: студенты, обучающиеся на контракте, менее активны на портале по сравнению с бюджетниками.

Что визуализировать: 
Различие в среднем числе взаимодействий между группами ("Name_OsnO") по следующим показателями: "s_all_avg", "s_course_viewed_avg", "s_q_attempt_viewed_avg".
Необходимо использовать в  plotly.express Boxplot распределения активности для бюджетников и контрактников.

```python
import pandas as pd
import plotly.express as px


# 2. Выбираем нужные столбцы для анализа
metrics = ['s_all_avg', 's_course_viewed_avg', 's_q_attempt_viewed_avg']


# 4. Расшифровываем форму обучения (1 - Бюджет, 2 - Контракт)
df['Основа_обучения'] = df['Name_OsnO'].map({1: 'Бюджет', 2: 'Контракт'})

# 5. Чтобы построить все 3 метрики на одном графике, используем функцию melt()
# Она перестроит таблицу так, чтобы названия метрик стали отдельным столбцом
df_melted = df.melt(
    id_vars=['Основа_обучения'], 
    value_vars=metrics, 
    var_name='Тип активности', 
    value_name='Среднее число взаимодействий'
)

# 6. Строим сгруппированный Boxplot
fig = px.box(
    df_melted,
    x='Тип активности',
    y='Среднее число взаимодействий',
    color='Основа_обучения',  # Разделяем цветом бюджетников и контрактников
    title='Сравнение активности: Бюджетники vs Контрактники (Задание 2)',
    labels={
        'Тип активности': 'Показатель активности',
        'Среднее число взаимодействий': 'Среднее количество',
        'Основа_обучения': 'Форма обучения'
    },
    color_discrete_map={'Бюджет': '#1f77b4', 'Контракт': '#ff7f0e'}, # Синий и оранжевый цвета
    template='plotly_white'
)

# 7. Немного приблизим график по оси Y, так как в таких данных часто бывают огромные выбросы
# Если хотите увидеть прямо все выбросы, закомментируйте эту строку
fig.update_yaxes(range=[0, df_melted['Среднее число взаимодействий'].quantile(0.95)]) 

# 8. Сохраняем результат в HTML файл вместо fig.show()
html_filename = "my_plot4.html"
fig.write_html(html_filename)

```

#### 1. Параметр s_all_avg (Среднее количество всех событий в неделю)

# Задание 3. Динамика активности студентов по неделям
Гипотеза: Активность студентов увеличивается перед сессией и уменьшается после неё.

Постройте линейный график изменения активности (s_all) по неделям (num_week).
Сравнить активность по курсам (kurs). Все курсы должны отображаться на одной диаграмме. 

```python
import pandas as pd
import plotly.express as px



# 1. Группируем данные по номеру недели и курсу.
# Считаем суммарную активность (sum). 
df_grouped = df.groupby(['num_week', 'Kurs'])['s_all'].mean().reset_index()

# 2. Превращаем номер курса в текст (например, "1 курс").
# Это нужно для того, чтобы Plotly воспринимал курс как отдельную категорию, 
# а не рисовал для него числовой цветовой градиент.
df_grouped['Kurs'] = df_grouped['Kurs'].astype(str) + ' курс'

# 3. Строим линейный график
fig = px.line(
    df_grouped,
    x='num_week',
    y='s_all',
    color='Kurs',
    title='Динамика суммарной активности студентов по неделям и курсам',
    labels={
        'num_week': 'Номер учебной недели',
        's_all': 'Общее количество действий (s_all)',
        'Kurs': 'Курс обучения'
    },
    markers=True,  # Добавляем точки на линиях, чтобы было видно каждую неделю
    template='plotly_white'
)

# 4. Настраиваем ось X, чтобы она показывала все недели без пропусков
fig.update_xaxes(dtick=1)

# 5. Сохраняем результат в HTML файл (чтобы обойти зависание fig.show())
html_file = "my_plot5.html"
fig.write_html(html_file)

```

Спокойное течение семестра (с 6 по 19 неделю): На протяжении большей части семестра графики всех курсов выглядят как относительно ровные горизонтальные линии. 

Приближение сессии (с 20 по 25 неделю): Начиная примерно с 19-20 учебной недели, на графике восходящий тренд графика абсолютно по всем курсам. 

График фиолетового цвета (5 курс) показываетмощный пик на 24-й неделе (более 50 действий в среднем). Это может быть связано с предзащитами дипломов (или что-то другое)

## Задание 4: Проанализировать динамику активности студентов с разным уровнем успеваемости (NameR_Level) в течение всего семестра.
Решите гипотезу: успеваемость студентов зависит от динамики их активности на образовательном портале в течение всего семестра. 

Построить линейный график (px.line), где:
X-ось — num_week (номер недели семестра)
Y-ось — s_all (общее количество событий)
Цветовая группировка — NameR_Level (успеваемость: отличники, хорошисты, троечники, двоечники)

```python
import pandas as pd
import plotly.express as px


# 1. Задаем словарь для расшифровки уровня успеваемости (NameR_Level)
level_mapping = {
    5: 'Отличники (5)',
    4: 'Хорошисты (4)',
    3: 'Троечники (3)',
    2: 'Двоечники (2)'
}

# Применяем расшифровку к новому столбцу, чтобы не сломать исходные данные
# Если в ваших данных есть другие значения (например, 0 или 1), они останутся NaN, 
# поэтому мы заполняем их как 'Без оценки' (по желанию)
df['Успеваемость'] = df['NameR_Level'].map(level_mapping).fillna('Без оценки/Зачет')

# 2. Группируем данные по неделе и успеваемости, считаем сумму событий s_all
df_grouped = df.groupby(['num_week', 'Успеваемость'])['s_all'].mean().reset_index()

# 3. Строим линейный график
fig = px.line(
    df_grouped,
    x='num_week',
    y='s_all',
    color='Успеваемость',
    title='Динамика активности студентов в зависимости от успеваемости',
    labels={
        'num_week': 'Номер учебной недели',
        's_all': 'Общее количество событий (s_all)',
        'Успеваемость': 'Уровень успеваемости'
    },
    markers=True, # Добавляем точки для наглядности каждой недели
    template='plotly_white',
    # Задаем порядок категорий в легенде для красоты (от отличников к двоечникам)
    category_orders={'Успеваемость': ['Отличники (5)', 'Хорошисты (4)', 'Троечники (3)', 'Двоечники (2)', 'Без оценки/Зачет']}
)

# 4. Настраиваем ось X, чтобы шаг был ровно в 1 неделю
fig.update_xaxes(dtick=1)

# 5. Сохраняем график в HTML файл
html_file = "my_plot6.html"
fig.write_html(html_file)

```

Четкая расслоение групп: На протяжении всего семестра графики не пересекаются и идут строго друг под другом. 
Самую высокую активность  демонстрируют отличники , за ними идут хорошисты , ниже — троечники  и на самая низкая активность у двоечников. 
Это доказывает, что высокая оценка — результат систематической работы, а не случайность.

Базовый (фоновый) уровень вовлеченности: В спокойный период семестра (с 8 по 18 неделю) отличники совершают в среднем 13–15 действий в неделю. В то же время студенты, получившие двойки, практически не заходят на портал (их базовый уровень всего 2–4 действия в неделю).

Характер подготовки к сессии: Отличники и хорошисты начинают наращивать активность заранее (уже с 19-й недели) и достигают высоких пиков (более 25 действий).

Троечники начинают готовиться немного позже, их график начинает резко расти только после 20-й недели, а пик активности смещен ближе к самому концу обучения (25-я неделя).

Двоечники даже в период сессии (пик на 24-25 неделе) не дотягивают даже до базового уровня отличников в середине семестра (максимум около 11-12 действий).

Гипотеза: Магистранты взаимодействуют с курсами иначе, чем бакалавры.

Построить воронку (метод go.Funnel()) активности студентов (s_course_viewed, s_q_attempt_viewed, s_a_submission_status_viewed) в зависимости от уровня образования (leveled).

```python
import pandas as pd
import plotly.graph_objects as go

# 1. Проверяем формат колонки LevelEd. Если там остались числа, переводим в текст:
if pd.api.types.is_numeric_dtype(df['LevelEd']):
    level_map = {1: "бакалавриат", 2: "магистратура", 3: "специалитет", 4: "аспирантура"}
    df['LevelEd'] = df['LevelEd'].replace(level_map)

# 2. Группируем по LevelEd и считаем средние
df_level = df.groupby("LevelEd")[[
    "s_course_viewed",
    "s_q_attempt_viewed",
    "s_a_submission_status_viewed"
]].mean().reset_index()

stages = ["Просмотры курса", "Просмотры тестов", "Отправка заданий"]

fig = go.Figure()

# 3. Перебираем все уровни образования, которые реально есть в сгруппированной таблице
for level_name in df_level["LevelEd"].unique():
    row = df_level[df_level["LevelEd"] == level_name]
    
    # Добавляем воронку для каждого найденного уровня
    fig.add_trace(go.Funnel(
        name=str(level_name).capitalize(),
        y=stages,
        x=row.iloc[:, 1:].values.flatten(),
        textinfo="value+percent initial" #Берем самое начальное значение воронки
    ))

fig.update_layout(
    title="Различие в активности бакалавров и магистров",
    template="plotly_white"
)

# Сохраняем в HTML
html_file = "my_plot7.html"
fig.write_html(html_file)

```

Стартовая активность (Просмотры курса): Студенты бакалавриата  на старте проявляют чуть более высокую активность: среднее число просмотров составляет 2.48 против 2.29 у магистрантов. Эти значения берутся за 100% базы.

Переход к тестам: На этапе просмотра тестов спад активности в обеих группах практически идентичен. Бакалавры сохраняют 69% от изначальной активности (показатель падает до 1.71), а магистранты — 71% (показатель падает до 1.63).

Отправка заданий: На финальном этапе, требующем максимальных усилий (отправка заданий на проверку), бакалавры оказываются немного более дисциплинированными. У них до этого этапа «доживает» 65% первоначальной активности, в то время как у магистрантов воронка сужается чуть сильнее — до 60%.

Общий итог: Характер взаимодействия с порталом у обеих групп идентичен — обе группы теряют около 30% активности при переходе от простого чтения к тестированию. Единственное заметное отличие: бакалавры в среднем совершают немного больше действий в абсолютных значениях на каждом из этапов.

## Задание 6. Различие в активности отличников, хорошистов, троечников и двоечников 
Гипотеза: Студенты с разным уровнем успеваемостью взаимодействуют с курсами иначе.

Построить воронку активности студентов (s_course_viewed, s_q_attempt_viewed, s_a_submission_status_viewed) в зависимости от уровня успеваемости (NameR_Level).

```python
import pandas as pd
import plotly.graph_objects as go



# 1. Словарь для перевода оценок в понятный текст
performance_map = {
    5: "Отличники (5)",
    4: "Хорошисты (4)",
    3: "Троечники (3)",
    2: "Двоечники (2)"
}

# Создаем новую колонку, чтобы не сломать исходные данные
if pd.api.types.is_numeric_dtype(df['NameR_Level']):
    df['Успеваемость'] = df['NameR_Level'].map(performance_map)
else:
    df['Успеваемость'] = df['NameR_Level']

# 2. Группируем по уровню успеваемости и считаем среднее количество действий
metrics = [
    "s_course_viewed",
    "s_q_attempt_viewed",
    "s_a_submission_status_viewed"
]
    
df_perf = df.groupby("Успеваемость")[metrics].mean().reset_index()

stages = ["Просмотры курса", "Просмотры тестов", "Отправка заданий"]

fig = go.Figure()

# 3. Задаем правильный порядок для вывода на график
ordered_levels = ["Отличники (5)", "Хорошисты (4)", "Троечники (3)", "Двоечники (2)"]

# Строим воронки в заданном порядке
for level_name in ordered_levels:
    row = df_perf[df_perf["Успеваемость"] == level_name]
    
    if not row.empty:
        fig.add_trace(go.Funnel(
            name=level_name,
            y=stages,
            x=row.iloc[:, 1:].values.flatten(),
            textinfo="value+percent initial" #Берем самое начальное значение воронки
        ))

fig.update_layout(
    title="Различие в активности студентов с разной успеваемостью",
    template="plotly_white"
)

# 4. Сохраняем график в HTML файл
html_file = "my_plot8.html"
fig.write_html(html_file)
```

 На первом этапе воронки («Просмотры курса») ВИДНА ЧЕТКАЯ ИЕРАРХИЯ. Отличники  заходят в материалы курса почти в 3 раза чаще (3.07 действий), чем двоечники (1.04 действия).

Стабильность отличников и хорошистов: Студенты  демонстрируют очень схожую, системную модель поведения. У обеих групп до финального этапа («Отправка заданий») сохраняется ровно 65% первоначальной активности. Переход от тестов к отправке заданий у них происходит почти без потерь.

Специфика троечников: Студенты с оценкой «3» имеют самый высокий процент просмотров тестов относительно старта (77%), но затем теряют вовлеченность, и до отправки заданий доходит лишь 62%.

Аномальное поведение двоечников: на этапе «Отправка заданий» их показатель внезапно возрастает до 59% ( значение 0.62 выше, чем 0.47 на предыдущем шаге).

## Задание 7:  Распределение студентов по кафедрам 
Определим распределение студентов по кафедрам, однаковое ли количество студентов?

Постройте горизонтальную столбчатую диаграмму (bar chart) для сравнения количества студентов на разных кафедрах.
Добавим интерактивность, цветовую палитру и подписи. На диаграмме должны быть выведены названия кафедр (возьмите данную информацию из Task 4.)

```python
import pandas as pd
import plotly.express as px
# Группируем основной датасет по идентификатору кафедры ('Depart').
# Для каждой кафедры считаем количество УНИКАЛЬНЫХ студентов ('userid').
df_dept = df.groupby('Depart')['userid'].nunique().reset_index()
df_dept.rename(columns={'userid': 'num_students'}, inplace=True)

# Загружаем справочник названий кафедр
df_deps = pd.read_csv('datasets/departments.csv')

#Объединяем логи со справочником по ID кафедры
df_merged = pd.merge(df_dept, df_deps, left_on='Depart', right_on='id', how='left')

# Сортируем по числу студентов (чтобы лидеры на графике были сверху)
df_merged = df_merged.sort_values(by='num_students', ascending=True)

# 4. Строим горизонтальный Bar Chart
fig = px.bar(
    df_merged,
    x='num_students',      
    y='name',              
    orientation='h',
    text='num_students',
    color='num_students',
    color_continuous_scale='Viridis',
    title='Распределение количества студентов по кафедрам',
    labels={
        'num_students': 'Количество студентов',
        'name': 'Кафедра'
    }
)

# Настраиваем высоту графика (чтобы вместились все 43 кафедры) и отступы
fig.update_layout(
    template='plotly_white',
    height=900,              
    yaxis_title=None,        
    margin=dict(l=100)       
)

# Выносим подписи со значениями за пределы столбцов
fig.update_traces(textposition='outside')


html_file = "my_plot9.html"
fig.write_html(html_file)
```

## Задание 8: Зависимость среднего балла от количества студентов на кафедре
Гипотеза: количество студентов, закрепленных на отдельной кафедре влияет на средний балл успеваемости по кафедрам.

Построить диаграмму рассевания по количеству студентов, среднему баллу, наванию кафедры. Параметр  size='num_students' - количество студентов. 

```python
import pandas as pd
import plotly.express as px


# Проверяем, что NameR_Level числовой (если в прошлых заданиях вы меняли его на текст)
if not pd.api.types.is_numeric_dtype(df['NameR_Level']):
    df['NameR_Level'] = pd.to_numeric(df['NameR_Level'], errors='coerce')

# Группируем по кафедре и считаем сразу две метрики
df_stats = df.groupby('Depart').agg(
    num_students=('userid', 'nunique'), # Считаем уникальных студентов
    mean_score=('NameR_Level', 'mean')  # Считаем средний балл
).reset_index()

# Загружаем файл с названиями кафедр и объединяем
df_deps = pd.read_csv('datasets/departments.csv')
df_merged = pd.merge(df_stats, df_deps, left_on='Depart', right_on='id', how='left')

# Округляем средний балл до 2 знаков после запятой для красоты
df_merged['mean_score'] = df_merged['mean_score'].round(2)

#Строим диаграмму рассеяния (Scatter plot)
fig = px.scatter(
    df_merged,
    x='num_students',          # По оси X - количество студентов
    y='mean_score',            # По оси Y - средний балл
    size='num_students',       # Размер пузырька зависит от количества студентов (по заданию)
    color='name',              # Раскрашиваем пузырьки по названиям кафедр
    hover_name='name',         # При наведении мышки будет крупно показываться название кафедры
    title='Зависимость среднего балла от количества студентов на кафедре',
    labels={
        'num_students': 'Количество студентов',
        'mean_score': 'Средний балл успеваемости',
        'name': 'Кафедра'
    },
    template='plotly_white',
    size_max=40                # Увеличиваем максимальный размер пузырька для наглядности
)


html_file = "my_plot10.html"
fig.write_html(html_file)
```

Маленькие кафедры (слева, до 50 человек): Здесь находятся точки с самыми высокими оценками (4.6, 4.9 и даже 5.0). Когда студентов мало, средний балл обычно выше.

Средние кафедры (в центре, от 100 до 250 человек): Это основная масса. Тут оценки очень разные — есть и низкие (3.5), и высокие (4.6).
